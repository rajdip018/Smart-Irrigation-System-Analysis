```python
import pandas as pd 
```

```python
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("irrigation_prediction.csv")
```

```python
df.head ()
```

Univariate Analysis

```python
#Temperature Distribution
```

```python
sns.histplot(df['Temperature_C'], kde=True)
plt.title("Temperature Distribution")
plt.show()
```

```python
#Soil Moisture Distribution
```

```python
sns.histplot(df['Soil_Moisture'], kde=True)
plt.title("Soil Moisture Distribution")
plt.show()
```

```python
#Irrigation Need
```

```python
sns.countplot(x='Irrigation_Need', data=df)
plt.title("Irrigation Need Count")
plt.show()
```

Bivariate Analysis

```python
#Temperature vs Soil Moisture
```

```python
sns.scatterplot(x='Temperature_C', y='Soil_Moisture', data=df)
plt.title("Temperature vs Soil Moisture")
plt.show()
```

```python
#Humidity vs Soil Moisture
```

```python
sns.scatterplot(x='Humidity', y='Soil_Moisture', data=df)
plt.show()
```

```python
#Soil Type vs Moisture
```

```python
sns.boxplot(x='Soil_Type', y='Soil_Moisture', data=df)
plt.show()
```

Multivariate Analysis

```python
#Correlation Heatmap
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

Season-Based Time Analysis

```python
df.columns
```

```python
sns.countplot(x='Season', hue='Irrigation_Need', data=df)
plt.title("Season vs Irrigation Need")
plt.show()
```

Category Analysis

```python
sns.countplot(x='Season', hue='Irrigation_Need', data=df)
plt.show()
```

Extract Key Insights

```python
print("Average Moisture:", df['Soil_Moisture'].mean())

print("Moisture when Temp > 30:")
print(df[df['Temperature_C'] > 30]['Soil_Moisture'].mean())
```

#Insights
1)High temperature significantly reduces soil moisture
2)Soil moisture below 30% indicates high irrigation need
3)Rainfall reduces dependency on irrigation
4)Sandy soil requires frequent irrigation
5)Seasonal variation impacts irrigation demand
